# Modele de KNN

```python
from sklearn.neighbors import BallTree
```

```python
# 1. On prépare les coordonnées (en radians pour la distance Haversine)
coords = np.deg2rad(df_final[['latitude', 'longitude']].values)

# 2. On crée l'arbre de décision (BallTree est très rapide pour les distances sphériques)
tree = BallTree(coords, metric='haversine')

# 3. Fonction pour trouver les structures les plus proches
def find_nearest(lat, lon, k=5):
    # Conversion de la position utilisateur en radians
    user_coord = np.deg2rad([[lat, lon]])
    
    # Recherche des k plus proches voisins
    distances, indices = tree.query(user_coord, k=k)
    
    # Conversion de la distance en kilomètres (rayon de la Terre ~6371km)
    distances_km = distances * 6371
    
    return df_final.iloc[indices[0]], distances_km[0]
```

# Créeons une fonction de Recommandation Intelligente

```python
def recommander_structures(lat, lon, categorie, k=3):
    # 1. Filtrer le dataset par catégorie choisie
    df_cat = df_final[df_final['Categorie'] == categorie].copy()
    
    if df_cat.empty:
        return "Aucune structure dans cette catégorie."

    # 2. Recréer un petit arbre pour cette catégorie (très rapide)
    coords_cat = np.deg2rad(df_cat[['latitude', 'longitude']].values)
    tree_cat = BallTree(coords_cat, metric='haversine')
    
    # 3. Chercher les k plus proches
    user_coord = np.deg2rad([[lat, lon]])
    dist, ind = tree_cat.query(user_coord, k=min(k, len(df_cat)))
    
    results = df_cat.iloc[ind[0]].copy()
    results['distance_km'] = dist[0] * 6371
    
    return results[['nom', 'Categorie', 'ville','quartier', 'telephone']]
```

# Test

```python
def test_caremap_model(user_lat, user_lon, category=None, k=3):
    """
    Teste le modèle KNN et affiche les structures les plus proches.
    """
    print(f"--- TEST DU MODÈLE CAREMAP ---")
    print(f"Position simulation : Lat {user_lat}, Lon {user_lon}")
    
    # Si une catégorie est spécifiée, on filtre d'abord
    if category:
        print(f"Filtre actif : {category}")
        df_test = df_final[df_final['Categorie'] == category].copy()
    else:
        df_test = df_final.copy()
    
    if df_test.empty:
        print("Erreur : Aucune donnée trouvée pour cette catégorie.")
        return

    # Reconstruction de l'index pour le sous-ensemble (très rapide)
    test_coords = np.deg2rad(df_test[['latitude', 'longitude']].values)
    test_tree = BallTree(test_coords, metric='haversine')
    
    # Calcul des voisins
    user_coord = np.deg2rad([[user_lat, user_lon]])
    dist, ind = test_tree.query(user_coord, k=min(k, len(df_test)))
    
    # Extraction et affichage
    results = df_test.iloc[ind[0]].copy()
    results['dist_km'] = dist[0] * 6371
    
    for i, (idx, row) in enumerate(results.iterrows()):
        print(f"{i+1}. {row['nom']}  ({row['Categorie']}) ({row['ville']}) ({row['quartier']})")
        print(f"   Distance : {row['dist_km']:.2f} km")
        print(f"   Contact : {row['telephone']}")
        print("-" * 30)

```

```python
import time

def benchmark_model(user_lat, user_lon, iterations=100):
    start_time = time.time()
    
    for _ in range(iterations):
        # On simule une requête de recherche
        user_coord = np.deg2rad([[user_lat, user_lon]])
        tree.query(user_coord, k=5)
        
    end_time = time.time()
    avg_time = (end_time - start_time) / iterations
    
    print(f"Temps moyen par requête : {avg_time * 1000:.4f} ms")
    return avg_time

# Test à Yaoundé
benchmark_model(3.8480, 11.5021)
```

Notre **moteur de recommandation** traite l'intégralité des structures de **santé** de **Yaoundé** et **Douala** en moins d'une **milliseconde**

```python
import streamlit as st
import pandas as pd
import numpy as np
import folium
from sklearn.neighbors import BallTree
from streamlit_folium import st_folium

# Configuration de la page
st.set_page_config(page_title="CareMap Cameroun", layout="wide")

# 1. Chargement des données (Assure-toi que le CSV est dans le même dossier)
@st.cache_data
def load_data():
    df = pd.read_csv('df_final_clean.csv') # Remplace par ton nom de fichier exact
    return df

try:
    df_final = load_data()

    # 2. Préparation du modèle global
    coords = np.deg2rad(df_final[['latitude', 'longitude']].values)
    tree = BallTree(coords, metric='haversine')

    st.title("📍 CareMap : Intelligence Sanitaire")

    # Sidebar pour les filtres
    st.sidebar.header("Options de recherche")
    categorie = st.sidebar.selectbox("Besoin de quoi ?", df_final['Categorie'].unique())
    k_voisins = st.sidebar.slider("Nombre de structures à trouver", 1, 10, 3)

    # Simulation de position (Bastos par défaut)
    st.sidebar.subheader("Ma Position")
    u_lat = st.sidebar.number_input("Latitude", value=3.9144, format="%.4f")
    u_lon = st.sidebar.number_input("Longitude", value=11.5841, format="%.4f")

    # 3. Logique de recommandation
    def get_recommendations(lat, lon, cat, k):
        df_cat = df_final[df_final['Categorie'] == cat].copy()
        if df_cat.empty:
            return None
        
        coords_cat = np.deg2rad(df_cat[['latitude', 'longitude']].values)
        tree_cat = BallTree(coords_cat, metric='haversine')
        
        dist, ind = tree_cat.query(np.deg2rad([[lat, lon]]), k=min(k, len(df_cat)))
        
        res = df_cat.iloc[ind[0]].copy()
        res['distance_km'] = dist[0] * 6371
        return res

    # 4. Affichage
    recos = get_recommendations(u_lat, u_lon, categorie, k_voisins)

    col1, col2 = st.columns([2, 1])

    with col1:
        # Création de la carte
        m = folium.Map(location=[u_lat, u_lon], zoom_start=13)
        folium.Marker([u_lat, u_lon], tooltip="Moi", icon=folium.Icon(color='blue')).add_to(m)

        if recos is not None:
            for _, row in recos.iterrows():
                folium.Marker(
                    [row['latitude'], row['longitude']],
                    popup=f"<b>{row['nom']}</b><br>{row['distance_km']:.2f} km",
                    icon=folium.Icon(color='red', icon='plus', prefix='fa')
                ).add_to(m)
        
        st_folium(m, width=800, height=500)

    with col2:
        st.subheader("Résultats")
        if recos is not None:
            for _, row in recos.iterrows():
                st.write(f"**{row['nom']}**")
                st.caption(f"📍 {row['quartier']} - 📞 {row['telephone']}")
                st.write(f"📏 Distance: {row['distance_km']:.2f} km")
                st.divider()

except Exception as e:
    st.error(f"Erreur au chargement : {e}")
    st.info("Vérifie que ton fichier CSV est bien présent et que les colonnes 'latitude', 'longitude' et 'Categorie' existent.")
```
